# Row-Magic Triangle (A396693)

This Jupyter notebook performs a complete enumeration of **row-magic triangles** of orders 1 through 6, a triangular generalization of magic squares in which the integers from 1 to \(n(n+1)/2\) are arranged in a triangular array such that all rows of the same length have equal sums, except for rows of lengths 1 and 2.

### Definition (OEIS A396693).
* a(n) is the number of order-n row-magic triangles formed by arranging the integers from 1 to n(n+1)/2 exactly once in a triangular array, such that, for every k from 2 to n, the sums of the three corner-anchored triangular subarrays of side length k are equal, counted up to rotations and reflections.

````python
import itertools
import numpy as np
````

## Orders considered in this notebook

### Order $n = 1$ (trivial case)

The order-1 Row-Magic Triangle is degenerate / trivial.

We record it here for completeness and do not treat it in detail.

```text
[1]
```

Thus, $a(1) = 1$.


### Order $n = 2$ (trivial case)

The order-2 Row-Magic Triangle is trivial, since no nontrivial row-sum constraints apply.

There is exactly one solution up to rotations and reflections:

```text
  1
 2 3
```

Unlike higher-order cases, the perimeter sums are not equal for $n = 2$, due to a limitation of the definition. The original author, Dairyong Jeong, defined the number of order-2 Row-Magic Triangles to be 1.


Thus, $a(2) = 1$.



### Order $n = 3$ (perimeter-magic case)

For $n = 3$, the row-magic condition requires that the sums of the three numbers along each side of the triangle be equal.

Thus, the order-3 Row-Magic Triangles coincide with the order-3 perimeter-magic triangles studied by Trotter.

There are exactly four distinct solutions up to rotations and reflections.

```text
    1
   6 4
  3 2 5
```

```text
    1
   6 5
  2 4 3
```

```text
    4
   3 2
  5 1 6
```

```text
    2
   5 3
  4 1 6
```

Thus, $a(3) = 4$.

### Order $n = 4$

Dairyong Jeong originally conjectured that no order-4 Row-Magic Triangles exist. However, this conjecture is incorrect.

There are exactly two distinct solutions up to rotations and reflections, and they are complementary to each other.

Thus, $a(4) = 2$.


```text
       1
     6   10
    5   4   8
  9   3   7   2
```

```text
      10
     5   1
    6   7   3
  2   8   4   9
```




````python

from itertools import permutations

# Configuration
RESTRICT_CORNERS = True  # True: require a > g > i
MAX_SAMPLES = 15

# Integers from 1 to 10
base = range(1, 11)


def check(p):
    a, b, c, d, e, f, g, h, k, i = p

    # Optional corner-order restriction
    if RESTRICT_CORNERS and not (a > g > i):
        return False

    # (1) Equal sums of the three inner rows
    t1 = b + e + k
    t2 = c + e + h
    t3 = d + e + f

    if not (t1 == t2 == t3):
        return False

    # (2) Equal sums of the three outer rows
    s1 = a + c + f + i
    s2 = a + b + d + g
    s3 = g + h + k + i

    return s1 == s2 == s3


def print_triangle(p):
    a, b, c, d, e, f, g, h, k, i = p

    print(f"      {a:2d}")
    print(f"    {b:2d}  {c:2d}")
    print(f"  {d:2d}  {e:2d}  {f:2d}")
    print(f"{g:2d}  {h:2d}  {k:2d}  {i:2d}")


# Enumerate all permutations
count = 0
samples = []

for perm in permutations(base):
    if check(perm):
        count += 1
        if len(samples) < MAX_SAMPLES:
            samples.append(perm)

# Display results
print(f"Total number of solutions: {count}\n")

if samples:
    print(f"Example solutions (up to {MAX_SAMPLES}):\n")
    for idx, sol in enumerate(samples, start=1):
        print(f"Solution {idx}:")
        print_triangle(sol)
        print()
````
