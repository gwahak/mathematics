### Trihexagonal Magic Figure (Order 6) – Solution Search Notebook

This Jupyter notebook searches for a valid magic labeling of the **order-6 trihexagonal figure**, a structure appearing in generalized magic-figure theory and related combinatorial designs.

This type of magic figure at order 6 is defined as an assignment of **distinct integers from 1 to 63** to the vertices such that:

- **Every triangular face** has constant sum **96**.
- **Every hexagonal face** has constant sum **192**.

The search terminates as soon as a valid labeling satisfying all constraints is found.

```python
ccode='''
#include <inttypes.h>
#include <stdbool.h>
#include <stdint.h>
#include <stdio.h>

/* 출력 순서 = 입력 Python 코드의 출력 순서. */
enum {
    aYY, aXX, aXY, a99, a9Y, a88, a89, a8X, a8Y,
    a77, a79, a7Y, a66, a67, a68, a69, a6X, a6Y,
    a55, a57, a59, a5Y, a44, a45, a46, a47, a48, a49, a4X, a4Y,
    a33, a35, a37, a39, a3Y, a22, a23, a24, a25, a26, a27, a28,
    a29, a2X, a2Y, a11, a13, a15, a17, a19, a1Y,
    a00, a01, a02, a03, a04, a05, a06, a07, a08, a09, a0X, a0Y,
    COUNT
};
_Static_assert(COUNT == 63, "Expected exactly 63 nodes");

static int v[COUNT];
static uint64_t solutions = 0;
static const uint64_t solution_limit = 1; /* 0: 끝까지 탐색 */
static const uint64_t FULL = (UINT64_C(1) << 63) - 1;

static bool put(uint64_t *used, int id, int value)
{
    if (value < 1 || value > 63) return false;
    uint64_t bit = UINT64_C(1) << (value - 1);
    if (*used & bit) return false;
    *used |= bit;
    v[id] = value;
    return true;
}

/* 각 단계에서 하나의 자유변수를 고르고, 나머지는 합 96으로 결정한다. */
#define P(id, expression) do { if (!put(used, (id), (expression))) return false; } while (0)
static bool assign_stage(int stage, int x, uint64_t *used)
{
    switch (stage) {
    case 1:
        P(a46, x); P(a57, 96 - v[a46] - v[a47]); break;
    case 2:
        P(a49, x);
        P(a59, 96 - v[a49] - v[a48]);
        P(a79, 96 - v[a46] - v[a49]); break;
    case 3:
        P(a68, x);
        P(a69, 96 - v[a68] - v[a79]);
        P(a67, 96 - v[a57] - v[a68]);
        P(a6X, 96 - v[a59] - v[a69]); break;
    case 4:
        P(a26, x);
        P(a27, 96 - v[a26] - v[a37]);
        P(a24, 96 - v[a57] - v[a27]);
        P(a29, 96 - v[a59] - v[a26]); break;
    case 5:
        P(a25, x);
        P(a35, 96 - v[a24] - v[a25]);
        P(a45, 96 - v[a35] - v[a46]);
        P(a15, 96 - v[a25] - v[a26]); break;
    case 6:
        P(a28, x);
        P(a39, 96 - v[a28] - v[a29]);
        P(a4X, 96 - v[a39] - v[a49]);
        P(a17, 96 - v[a27] - v[a28]); break;
    case 7:
        P(a89, x); P(a8X, 96 - v[a89] - v[a79]); break;
    case 8:
        P(a77, x);
        P(a66, 96 - v[a77] - v[a67]);
        P(a44, 96 - v[a47] - v[a77]);
        P(a55, 96 - v[a44] - v[a45]);
        P(a22, 96 - v[a25] - v[a55]);
        P(a99, 96 - v[a22] - v[a29]);
        P(a88, 96 - v[a99] - v[a89]); break;
    case 9:
        P(a7Y, x);
        P(a6Y, 96 - v[a7Y] - v[a6X]);
        P(aYY, 96 - v[a6Y] - v[a66]);
        P(a4Y, 96 - v[a48] - v[a7Y]);
        P(a5Y, 96 - v[a4Y] - v[a4X]);
        P(a2Y, 96 - v[a28] - v[a5Y]);
        P(a9Y, 96 - v[a24] - v[a2Y]);
        P(a8Y, 96 - v[a9Y] - v[a8X]); break;
    case 10:
        P(a23, x);
        P(a33, 96 - v[a22] - v[a23]);
        P(a13, 96 - v[a23] - v[a24]); break;
    case 11:
        P(a00, x);
        P(a03, 96 - v[a00] - v[a33]);
        P(a05, 96 - v[a00] - v[a55]);
        P(a07, 96 - v[a00] - v[a77]);
        P(a09, 96 - v[a00] - v[a99]);
        P(a0Y, 96 - v[a00] - v[aYY]);
        P(a02, 96 - v[a13] - v[a03]);
        P(a04, 96 - v[a15] - v[a05]);
        P(a06, 96 - v[a17] - v[a07]); break;
    case 12:
        P(a19, x);
        P(a08, 96 - v[a19] - v[a09]);
        P(a2X, 96 - v[a19] - v[a29]);
        P(a3Y, 96 - v[a2X] - v[a2Y]); break;
    case 13:
        P(aXX, x); P(aXY, 96 - v[aXX] - v[aYY]);
        if (v[aXX] <= v[aXY]) return false;
        break;
    case 14:
        P(a0X, x); P(a1Y, 96 - v[a0X] - v[a0Y]);
        if (v[a0X] <= v[a1Y]) return false;
        break;
    case 15:
        P(a01, x); P(a11, 96 - v[a01] - v[a00]);
        if (v[a01] <= v[a11]) return false;
        break;
    default:
        return false;
    }
    return true;
}
#undef P

static bool search(int stage, uint64_t used)
{
    if (stage == 16) {
        if (used != FULL) return false;
        ++solutions;
        printf("%" PRIu64 "th: solution:", solutions);
        for (int j = 0; j < COUNT; ++j) printf(" %d", v[j]);
        putchar('\n');
        return solution_limit != 0 && solutions >= solution_limit;
    }
    for (int x = 1; x <= 63; ++x) {
        if (used & (UINT64_C(1) << (x - 1))) continue;
        uint64_t next_used = used;
        if (assign_stage(stage, x, &next_used) &&
            search(stage + 1, next_used)) return true;
    }
    return false;
}

int main(void)
{
    for (int x = 33; x <= 63; ++x) {
        uint64_t base = 0;
        if (!put(&base, a47, x)) continue;

        /* range((96-a47)//2+1, min(a47,96-a47)) */
        for (int y = (96 - x) / 2 + 1; y < x && y < 96 - x; ++y) {
            uint64_t used = base;

            if (!put(&used, a48, y)) continue;
            if (!put(&used, a37, 96 - y - x)) continue;

            printf("a47 = %d, a48 = %d\n", x, y);
            fflush(stdout);

            if (search(1, used)) goto done;
        }
    }

done:
    printf("total: %" PRIu64 "\n", solutions);
    return 0;
}'''
print(ccode)
```

```python
import matplotlib.pyplot as plt

# 위쪽에 표시할 숫자 값
rows2 = [
[33],
[40, 23],
[41, -1, 35],
[4, 51, 2, 59],
[52, -1, 43, -1, 11],
[7, 37, 5, 48, 29, 56],
[60, -1, 54, -1, 19, -1, 42],
[10, 26, 8, 34, 32, 45, 1, 53],
[14, -1, 62, -1, 30, -1, 50, -1, 13],
[24, 58, 22, 12, 46, 20, 15, 31, 44, 39],
[6, -1, 16, -1, 38, -1, 61, -1, 21, -1, 3],
[27, 63, 25, 55, 49, 9, 18, 17, 47, 28, 57, 36],
]

# 원래 변수 구조 (필요하면 좌표 확인용으로 유지 가능)
rows = [
    ["aYY"],
    ["aXX", "aXY"],
    ["a99", -1, "a9Y"],
    ["a88", "a89", "a8X", "a8Y"],
    ["a77", -1, "a79", -1, "a7Y"],
    ["a66", "a67", "a68", "a69", "a6X", "a6Y"],
    ["a55", -1, "a57", -1, "a59", -1, "a5Y"],
    ["a44", "a45", "a46", "a47", "a48", "a49", "a4X", "a4Y"],
    ["a33", -1, "a35", -1, "a37", -1, "a39", -1, "a3Y"],
    ["a22", "a23", "a24", "a25", "a26", "a27", "a28", "a29", "a2X", "a2Y"],
    ["a11", -1, "a13", -1, "a15", -1, "a17", -1, "a19", -1, "a1Y"],
    ["a00", "a01", "a02", "a03", "a04", "a05", "a06", "a07", "a08", "a09", "a0X", "a0Y"],
]


rows_order=rows2
radius = 0.9
x_spacing = 2.2
y_spacing = 1.9

fig = plt.figure(figsize=(9, 9))
ax = plt.gca()
ax.set_aspect("equal")
ax.axis("off")

max_len = max(len(r) for r in rows)
xs, ys = [], []

for ridx, row in enumerate(rows):
    offset = (max_len - len(row)) * x_spacing / 2

    for cidx, val in enumerate(row):
        if val == -1:
            continue

        x = offset + cidx * x_spacing
        y = -ridx * y_spacing
        xs.append(x)
        ys.append(y)

        # 바깥 원
        circle = plt.Circle(
            (x, y), radius,
            fill=False, linewidth=1.2, edgecolor="black"
        )
        ax.add_patch(circle)

        # 가운데 가로선 (항상 검정)
        ax.plot(
            [x - radius * 0.85, x + radius * 0.85],
            [y, y],
            linewidth=1.0,
            color="black"
        )

        # 위쪽: 순서 라벨 혹은 숫자
        order_val = rows_order[ridx][cidx]
        if order_val != -1 and order_val != "":
            ax.text(
                x, y + radius * 0.35,
                str(order_val),
                ha='center', va='center',
                fontsize=11, color='black'
            )

        # 아래쪽: 이름
        ax.text(
            x, y - radius * 0.35,
            str(val),
            ha='center', va='center',
            fontsize=11, color='black'
        )

margin = 1.5
ax.set_xlim(min(xs) - margin, max(xs) + margin)
ax.set_ylim(min(ys) - margin, max(ys) + margin)

plt.tight_layout()
plt.show()
```
